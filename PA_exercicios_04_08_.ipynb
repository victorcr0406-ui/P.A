Exercício 6 – Situação prática
Um produto custa R$ 80,00. O cliente ganhou 10% de desconto.
Faça um programa que calcule e mostre:
• O valor do desconto
• O valor final a pagar
• Se o valor final é menor que 75.00 (True ou False)

```python
preco_original = 80.00
percentual_desconto = 0.10
valor_desconto = preco_original * percentual_desconto
valor_final = preco_original - valor_desconto
e_menor_que_75 = valor_final < 75.00
print(f"Valor do desconto: R$ {valor_desconto:.2f}")
print(f"Valor final a pagar: R$ {valor_final:.2f}")
print(f"O valor final é menor que 75.00? {e_menor_que_75}")
```
