Exercício 1 — Contagem

Faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while.

```python

numero_limite = int(input("Digite um número inteiro: "))

contador = 1

while contador <= numero_limite:
    print(contador)
    contador += 1
```

Exercício 2 — Contagem regressiva

Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

```python
numero = int(input("Digite um número inicial para a contagem regressiva: "))

while numero >= 0:
    print(numero)
    numero -= 1
```

Exercício 4 — Soma dos números

Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número.

```python
numero_limite = int(input("Digite um número inteiro: "))

contador = 1
soma = 0

while contador <= numero_limite:
    soma += contador
    contador += 1

print(f"A soma de 1 até {numero_limite} é: {soma}")
```

Exercício 5 — Tabuada

Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.

```python

numero = int(input("Digite um número para ver sua tabuada: "))

contador = 1

print(f"\nTabuada do {numero}:")

while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador += 1
```

Exercício 6 — Senha

Crie um programa que peça uma senha ao usuário.

Enquanto a senha estiver incorreta, mostre:

Senha incorreta!

Quando o usuário digitar a senha correta, mostre:

Acesso permitido!

Considere que a senha correta seja "1234".

```python

senha_correta = "1234"

senha_digitada = input("Digite a senha: ")

while senha_digitada != senha_correta:
    print("Senha incorreta!")
    senha_digitada = input("Digite a senha novamente: ")

print("Acesso permitido!")
```

Exercício 7 — Somando até digitar zero

Faça um programa que peça vários números ao usuário.

O programa deve continuar pedindo números até que o usuário digite 0.

Ao final, mostre a soma de todos os números digitados.

```python
soma = 0

numero = int(input("Digite um número (ou 0 para sair): "))

while numero != 0:
    soma += numero
    numero = int(input("Digite outro número (ou 0 para sair): "))

print(f"A soma de todos os números digitados é: {soma}")
```

Exercício 8 — Maior número

Peça números ao usuário até que ele digite 0.

Ao final, mostre qual foi o maior número digitado

```python

numero = int(input("Digite um número (ou 0 para sair): "))

maior = numero

while numero != 0:
    if numero > maior:
        maior = numero
    numero = int(input("Digite outro número (ou 0 para sair): "))

if maior == 0:
    print("Nenhum número diferente de zero foi digitado.")
else:
    print(f"O maior número digitado foi: {maior}")
```

Exercício 9 — Média das notas

Faça um programa que peça 5 notas de um aluno usando while.

Ao final, calcule e mostre a média das notas.

```python

soma_notas = 0
contador = 1

while contador <= 5:
    nota = float(input(f"Digite a {contador}ª nota: "))
    soma_notas += nota
    contador += 1

media = soma_notas / 5

print(f"\nA média final do aluno é: {media:.2f}")
```

Exercício 10 — Caixa eletrônico

Faça um programa que comece com um saldo de R$ 1000,00.

Mostre um menu:

1 - Depositar

2 - Sacar

3 - Ver saldo

4 - Sai

```python

saldo = 1000.0

opcao = 0

while opcao != 4:
    print("\n--- CAIXA ELETRÔNICO ---")
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        valor_deposito = float(input("Digite o valor para depósito: R$ "))
        if valor_deposito > 0:
            saldo += valor_deposito
            print(f"Depósito de R$ {valor_deposito:.2f} realizado com sucesso!")
        else:
            print("Valor inválido para depósito.")

    elif opcao == 2:
        valor_saque = float(input("Digite o valor para saque: R$ "))
        if valor_saque <= 0:
            print("Valor inválido para saque.")
        elif valor_saque <= saldo:
            saldo -= valor_saque
            print(f"Saque de R$ {valor_saque:.2f} realizado com sucesso!")
        else:
            print("Saldo insuficiente para realizar o saque!")

    elif opcao == 3:
        print(f"Seu saldo atual é: R$ {saldo:.2f}")

    elif opcao == 4:
        print("Obrigado por utilizar nosso caixa eletrônico. Até logo!")

    else:
        print("Opção inválida! Tente novamente.")
```
