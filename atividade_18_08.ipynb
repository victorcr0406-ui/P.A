Exercício 1 – Idade para dirigir
Receba a idade de uma pessoa e diga se ela pode tirar carteira de motorista. A pessoa
pode tirar a carteira se tiver idade maior ou igual a 18 anos.

```python
idade = int(input("Digite a sua idade: "))

if idade >= 18:
    print("Você pode tirar a carteira de motorista! 🚗")
else:
    print("Você ainda não pode tirar a carteira de motorista.")
```

Exercício 2 – Par ou Ímpar
Receba um número e diga se ele é par ou ímpar.

```python
numero = int(input("Digite um número inteiro: "))

if numero % 2 == 0:
    print(f"O número {numero} é PAR! ")
else:
    print(f"O número {numero} é ÍMPAR! ")
```

Exercício 3 – Maior de dois números
Receba dois números e mostre qual é o maior. Se os dois números forem iguais, diga que
eles são iguais.

Exercício 5 – Desconto na compra
Receba o valor de uma compra. Se o valor for maior ou igual a 100, aplique 10% de
desconto e mostre o valor final. Caso contrário, mostre o valor normal da compra.

```python
valor_compra = float(input("Digite o valor da compra (R$): "))

if valor_compra >= 100:
    desconto = valor_compra * 0.10
    valor_final = valor_compra - desconto

    print("Você ganhou 10% de desconto!")
    print(f"Valor original: R$ {valor_compra:.2f}")
    print(f"Desconto: R$ {desconto:.2f}")
    print(f"Valor final a pagar: R$ {valor_final:.2f}")
else:
    print("Sua compra não atingiu o valor mínimo para desconto.")
    print(f"Valor total a pagar: R$ {valor_compra:.2f}")
```

Exercício 7 – Classificação de idade
Receba a idade de uma pessoa e classifique-a de acordo com sua faixa etária:
 Menor de 12 anos → Criança
 De 12 a 17 anos → Adolescente
 De 18 a 59 anos → Adulto
 60 anos ou mais → Idoso

```python
idade = int(input("Digite a sua idade: "))

if idade < 12:
    print("Classificação: Criança ")
elif idade <= 17:
    print("Classificação: Adolescente ")
elif idade <= 59:
    print("Classificação: Adulto ")
else:
    print("Classificação: Idoso ")
```

Exercício 8 – Média de duas notas
Receba duas notas, calcule a média e diga se o aluno foi aprovado ou reprovado. O aluno
será aprovado se a média for maior ou igual a 7.

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

print(f"Média: {media:.1f}")

if media >= 7:
    print("Situação: Aprovado! ")
else:
    print("Situação: Reprovado. ")
```

Exercício 10 – Calculadora de IMC
Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
IMC = peso / (altura × altura)
Depois, classifique o resultado:
 IMC menor que 18.5 → Abaixo do peso
 IMC de 18.5 a 24.9 → Peso normal
 IMC de 25 a 29.9 → Sobrepeso
 IMC maior ou igual a 30 → Obesidad

```python
peso = float(input("Digite o seu peso em kg: "))
altura = float(input("Digite a sua altura em metros: "))

imc = peso / (altura * altura)

print(f"Seu IMC é: {imc:.2f}")

if imc < 18.5:
    print("Classificação: Abaixo do peso ")
elif imc <= 24.9:
    print("Classificação: Peso normal ")
elif imc <= 29.9:
    print("Classificação: Sobrepeso ")
else:
    print("Classificação: Obesidade ")
```
